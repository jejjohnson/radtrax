# Plume vetting: a spectral test for matched-filter candidates

The matched filter scores each pixel by one projection of its spectrum, so anything that removes light under methane's 2.3 μm band scores: carbonate soils, salt pans, shadows, striping. Plume vetting (Xiang et al. 2024; Růžička et al. 2026) goes back to the radiances. It pairs a candidate's strongest pixels with pixels outside it that have the same surface, divides them, and fits methane's transmittance times a smooth continuum to the ratio; the misfit $D_{\text{norm}}$ accepts or rejects the candidate. This notebook builds one EMIT-like scene in which the matched filter scores a carbonate patch above a real plume, and vets both.

**What you'll learn:**

1. How to get the transmittance the fit needs from [`ForwardModel`](xref:api#radtrax.ForwardModel), and why a constant absorption coefficient biases the column low.
2. How seeds, one-to-one pairing and the ratio spectrum cancel the surface.
3. How the profile fit and $D_{\text{norm}}$ separate a plume from a confounder, and how a null from the same scene calibrates the score.

The vetting kernel does not exist in radtrax yet. ## Background

For a pixel under a plume and a pixel outside it with the same surface, sun and atmosphere, the ratio of their radiances cancels everything but the plume and what the two surfaces do not share:

$$R(\lambda) = \frac{\overline{L_{\text{in}}}(\lambda)}{\overline{L_{\text{out}}}(\lambda)} \approx P(\lambda; c)\, E(\lambda; \Delta X), \qquad E_b(\Delta X) = \frac{y_b(\text{state} + \Delta X)}{y_b(\text{state})}$$

with $E$ methane's band transmittance for a column enhancement $\Delta X$ [mol/m²] and $P$ a low-order Legendre series that absorbs the surface mismatch. The fit gives $\widehat{\Delta X}$, and the score divides out the continuum, $q = R / P$, and compares $q$ with $E$ relative to $q$'s own spread:

$$D_{\text{norm}} = \frac{\operatorname{mean}_\lambda \lvert q - E(\widehat{\Delta X}) \rvert}{\operatorname{mean}_\lambda \lvert q - \bar q \rvert}$$

A perfect fit gives 0; no methane gives $\widehat{\Delta X} = 0$, $E = 1$ and $D_{\text{norm}} \approx 1$. EMIT's daily digest alerts below 0.3.

## 1. The transmittance the fit needs

Nine EMIT-like bands (8.5 nm wide, every 7.5 nm, 2296–2356 nm) on *radtrax · 3*'s synthetic table of 150 Voigt lines, so the notebook runs without HITRAN. The model is AFGL US Standard to 20 km with CH₄ and a stand-in H₂O, a 35° sun and a 10° view. `transmittance_table` puts a column enhancement $\Delta X$ in the bottom layer and divides by the plume-free radiance; `linear=True` gives the reference code's older model, $\exp(-\varepsilon \Delta X)$ with $\varepsilon = -\partial \ln y / \partial \Delta X$ at 0, which is the matched filter's target from [`unit_absorption_target`](xref:api#radtrax.unit_absorption_target).

```python
CENTERS_NM = np.arange(2296.0, 2357.0, 7.5)  # (B,) = (9,)
opacity = TableOpacity.from_datasets(
    {"CH4": synthetic_table(seed=0), "H2O": synthetic_table(seed=1)}
)
instrument = GaussianInstrument(NU_CM, CENTERS_NM, np.full(CENTERS_NM.size, 8.5))
model = radtrax.ForwardModel(opacity, BeerLambert(), instrument)
state = afgl_us_standard(z_top_km=20.0, gases=["CH4", "H2O"])
geometry = ViewingGeometry.from_angles(35.0, 10.0)
DELTA_GRID = np.linspace(0.0, 0.4, 801)  # ΔX [mol/m²]                         (D,)

# 1 ppm·m at the surface layer, in mol/m², for quoting columns as EMIT does
n_air = float(state.p_pa[0] / (BOLTZMANN_J_PER_K * state.t_k[0]))  # molecules/m³
PPMM = 1e-6 * n_air / AVOGADRO_PER_MOL
print(f"1 ppm·m = {PPMM:.3e} mol/m² at the surface layer")
```

## 2. The vetting kernel, before it lands in radtrax

Each definition below is the reference implementation in the issue that will add it to `radtrax.retrieval.vetting`: [#154](https://github.com/jejjohnson/radtrax/issues/154) `transmittance_table`, [#155](https://github.com/jejjohnson/radtrax/issues/155) `VettingWindow`, [#148](https://github.com/jejjohnson/radtrax/issues/148) `select_seeds` and `background_pool`, [#149](https://github.com/jejjohnson/radtrax/issues/149) `pair_pixels` and `ratio_spectrum`, [#150](https://github.com/jejjohnson/radtrax/issues/150) `fit_transmittance` and `d_norm`, [#151](https://github.com/jejjohnson/radtrax/issues/151) `vet_candidate`, [#152](https://github.com/jejjohnson/radtrax/issues/152) `null_masks` and `null_pvalue`. The names, arguments and shapes are the planned public API; when an issue lands, its block here becomes an import.

```python
# #154 transmittance_table, #155 VettingWindow


def transmittance_table(
    model: radtrax.ForwardModel,
    state,
    geometry: ViewingGeometry,
    delta_grid: Float[Array, " D"],
    *,
    gas: str = "CH4",
    linear: bool = False,
) -> Float[Array, "D B"]:
    """E_b(ΔX) = y_b(state + ΔX) / y_b(state) on a grid of enhancements [mol/m²]."""
    delta_grid = jnp.asarray(delta_grid)
    if linear:
        t = radtrax.unit_absorption_target(
            model, state, Lambertian(1.0), geometry, gas=gas
        )
        return jnp.exp(jnp.outer(delta_grid, t))  # t = ∂ ln y/∂ΔX = −ε

    def y(delta):
        return model(state.add_column(gas, delta), Lambertian(1.0), geometry)

    y_all = jax.vmap(y)(delta_grid)  # (D, B)
    return y_all / y(0.0)


@dataclasses.dataclass(frozen=True)
class VettingWindow:
    """A gas's fitting bands, the bands kept out of pairing, and the pool cut."""

    gas: str
    fit_nm: tuple[tuple[float, float], ...]
    absorbing_nm: tuple[tuple[float, float], ...]
    mf_background_mol_m2: float

    @classmethod
    def ch4_emit(cls) -> "VettingWindow":
        """EMIT's CH₄ window, as the reference code's configuration sets it."""
        ppmm = 4.03e-5  # mol/m² per ppm·m at the surface
        return cls(
            "CH4",
            ((2100.0, 2440.0),),
            ((1640.0, 1690.0), (2100.0, 2440.0)),
            30.0 * ppmm,
        )

    def fit_bands(self, wavelength_nm) -> np.ndarray:
        wl = np.asarray(wavelength_nm)
        return np.any([(wl >= a) & (wl <= b) for a, b in self.fit_nm], axis=0)

    def pairing_bands(self, wavelength_nm) -> np.ndarray:
        wl = np.asarray(wavelength_nm)
        inside = np.any([(wl >= a) & (wl <= b) for a, b in self.absorbing_nm], axis=0)
        return ~inside
```

```python
# #148 select_seeds, background_pool; #149 pair_pixels, ratio_spectrum


def _grow(mask: Bool[Array, "H W"]) -> Bool[Array, "H W"]:
    """One 3 × 3 dilation, as a max-pool."""
    pooled = jax.lax.reduce_window(
        mask.astype(float), 0.0, jax.lax.max, (3, 3), (1, 1), "SAME"
    )
    return pooled > 0


def select_seeds(
    mf: Float[Array, "H W"],
    mask: Bool[Array, "H W"],
    *,
    k: int = 40,
    cap_percentile: float = 99.0,
    grow: int = 1,
) -> Bool[Array, "H W"]:
    """The k strongest matched-filter pixels in mask below its cap, grown inside it."""
    mf, mask = jnp.asarray(mf), jnp.asarray(mask)
    cap = jnp.nanpercentile(jnp.where(mask, mf, jnp.nan), cap_percentile)
    score = jnp.where(mask & (mf <= cap), mf, -jnp.inf)
    kth = jnp.sort(score.ravel())[-k]
    seeds = jnp.isfinite(score) & (score >= kth)
    for _ in range(grow):
        seeds = _grow(seeds) & mask
    return seeds


def background_pool(
    mf: Float[Array, "H W"],
    seeds: Bool[Array, "H W"],
    *,
    threshold: float,
    exclude: Bool[Array, "C H W"] | None = None,
    box: int = 200,
) -> Bool[Array, "H W"]:
    """Pixels within box of the seeds' bounding box, outside every candidate, |MF| small."""
    rows, cols = jnp.nonzero(seeds)
    i = jnp.arange(mf.shape[0])[:, None]
    j = jnp.arange(mf.shape[1])[None, :]
    near = (i >= rows.min() - box) & (i <= rows.max() + box)
    near &= (j >= cols.min() - box) & (j <= cols.max() + box)
    pool = near & ~seeds & (jnp.abs(mf) < threshold)
    if exclude is not None:
        pool &= ~jnp.any(jnp.asarray(exclude), axis=0)
    return pool


def pair_pixels(
    cube: Float[Array, "H W B"],
    seeds: Bool[Array, "H W"],
    pool: Bool[Array, "H W"],
    pairing_bands: Bool[Array, " B"],
    *,
    keep: float = 0.5,
) -> tuple[Int[np.ndarray, " N"], Int[np.ndarray, " N"]]:
    """One-to-one pairs minimising the Euclidean distance on pairing bands; best kept."""
    flat = np.asarray(cube).reshape(-1, cube.shape[-1])
    i_in = np.flatnonzero(np.asarray(seeds))
    i_out = np.flatnonzero(np.asarray(pool))
    s_in = flat[i_in][:, pairing_bands]
    s_out = flat[i_out][:, pairing_bands]
    d = np.linalg.norm(s_in[:, None] - s_out[None], axis=-1)  # (n_in, n_pool)
    rows, cols = linear_sum_assignment(d)
    best = np.argsort(d[rows, cols])[: int(keep * rows.size)]
    return i_in[rows[best]], i_out[cols[best]]


def ratio_spectrum(
    cube: Float[Array, "H W B"],
    in_idx: Int[np.ndarray, " N"],
    out_idx: Int[np.ndarray, " N"],
) -> Float[Array, " B"]:
    """R = mean L_in / mean L_out over the pairs."""
    flat = jnp.asarray(cube).reshape(-1, cube.shape[-1])
    return flat[in_idx].mean(0) / flat[out_idx].mean(0)
```

```python
# #150 TransmittanceFit, fit_transmittance, d_norm


class TransmittanceFit(eqx.Module):
    """The profile fit of R ≈ P(λ; c) E(λ; ΔX)."""

    delta: Float[Array, ""]  # ΔX̂ [mol/m²]
    coefficients: Float[Array, " K"]
    continuum: Float[Array, " B"]  # P(λ; ĉ)
    transmittance: Float[Array, " B"]  # E(λ; ΔX̂)
    residual_ss: Float[Array, ""]


def fit_transmittance(
    ratio: Float[Array, " B"],
    table: Float[Array, "D B"],
    delta_grid: Float[Array, " D"],
    wavelength_nm: Float[Array, " B"],
    *,
    degree: int = 2,
) -> TransmittanceFit:
    """Closed-form least squares for c at each ΔX, then the ΔX of least residual."""
    wl = jnp.asarray(wavelength_nm)
    x = 2 * (wl - wl.min()) / (wl.max() - wl.min()) - 1
    V = legendre_polynomials(x, degree)  # (B, K)
    A = jnp.asarray(table)[:, :, None] * V[None]  # (D, B, K)
    c = jax.vmap(lambda a: jnp.linalg.lstsq(a, ratio)[0])(A)  # (D, K)
    sse = jnp.sum((jnp.einsum("dbk,dk->db", A, c) - ratio) ** 2, axis=1)  # (D,)
    d = jnp.argmin(sse)
    return TransmittanceFit(
        jnp.asarray(delta_grid)[d], c[d], V @ c[d], jnp.asarray(table)[d], sse[d]
    )


def d_norm(ratio: Float[Array, " B"], fit: TransmittanceFit) -> Float[Array, ""]:
    """Xiang et al. eq. 11: mean|q − E| / mean|q − q̄|, q = R / P."""
    q = ratio / fit.continuum
    return jnp.mean(jnp.abs(q - fit.transmittance)) / jnp.mean(jnp.abs(q - q.mean()))
```

```python
# #151 VettingResult, vet_candidate; #152 null_masks, null_pvalue


class VettingResult(eqx.Module):
    """One candidate's scores and the spectra behind them."""

    delta: Float[Array, ""]
    d_norm: Float[Array, ""]
    n_pairs: int = eqx.field(static=True)
    ratio: Float[Array, " B"]
    fit: TransmittanceFit


def vet_candidate(
    cube: Float[Array, "H W B"],
    mf: Float[Array, "H W"],
    mask: Bool[Array, "H W"],
    wavelength_nm: Float[Array, " B"],
    table: Float[Array, "D Bf"],
    delta_grid: Float[Array, " D"],
    window: VettingWindow,
    *,
    exclude: Bool[Array, "C H W"] | None = None,
    k: int = 40,
    keep: float = 0.5,
    degree: int = 2,
    box: int = 200,
) -> VettingResult:
    """Seeds, pool, pairs, ratio, fit and D_norm for one candidate mask."""
    fit_b = window.fit_bands(wavelength_nm)
    seeds = select_seeds(mf, mask, k=k)
    pool = background_pool(
        mf, seeds, threshold=window.mf_background_mol_m2, exclude=exclude, box=box
    )
    i_in, i_out = pair_pixels(
        cube, seeds, pool, window.pairing_bands(wavelength_nm), keep=keep
    )
    ratio = ratio_spectrum(cube, i_in, i_out)[fit_b]
    fit = fit_transmittance(
        ratio, table, delta_grid, np.asarray(wavelength_nm)[fit_b], degree=degree
    )
    return VettingResult(fit.delta, d_norm(ratio, fit), int(i_in.size), ratio, fit)


def null_masks(
    mask: Bool[Array, "H W"],
    n: int,
    *,
    key: Array,
    exclude: Bool[Array, "C H W"] | None = None,
    min_fraction: float = 0.9,
) -> Bool[np.ndarray, "S H W"]:
    """n copies of mask, rotated and translated inside the scene, off every candidate."""
    mask = np.asarray(mask)
    banned = np.zeros_like(mask) if exclude is None else np.any(exclude, axis=0)
    H, W = mask.shape
    out = []
    while len(out) < n:
        key, k_angle, k_row, k_col = jax.random.split(key, 4)
        angle = float(jax.random.uniform(k_angle, maxval=360.0))
        m = rotate(mask.astype(float), angle, reshape=False, order=0) > 0.5
        r, c = np.nonzero(m)
        if r.size < min_fraction * mask.sum():
            continue
        dr = int(jax.random.randint(k_row, (), -r.min(), H - r.max()))
        dc = int(jax.random.randint(k_col, (), -c.min(), W - c.max()))
        m = shift(m.astype(float), (dr, dc), order=0) > 0.5
        if m.sum() >= min_fraction * mask.sum() and not (m & banned).any():
            out.append(m)
    return np.stack(out)


def null_pvalue(score: float, null_scores: Float[Array, " S"]) -> float:
    """p̂ = mean 1[D_norm(copy) ≤ D_norm(candidate)]."""
    return float(jnp.mean(jnp.asarray(null_scores) <= score))
```

With the kernel in hand, the transmittance table and the constant-coefficient check. A fit of the exact transmittance with a constant coefficient returns too small a column, because lines saturate: a strong plume absorbs less per molecule than the weak-absorption slope predicts.

```python
E_TABLE = transmittance_table(model, state, geometry, DELTA_GRID)  # (D, B)
E_LIN = transmittance_table(model, state, geometry, DELTA_GRID, linear=True)
eps = -np.log(np.asarray(E_LIN[1])) / DELTA_GRID[1]  # ε [m²/mol]           (B,)
for dx in (0.05, 0.1, 0.3):
    E_exact = np.asarray(E_TABLE[np.argmin(np.abs(DELTA_GRID - dx))])
    dx_lin = -(eps @ np.log(E_exact)) / (eps @ eps)
    print(
        f"ΔX = {dx:.2f} mol/m² ({dx / PPMM:,.0f} ppm·m): constant ε gives "
        f"{100 * (dx_lin / dx - 1):+.1f} %"
    )
deepest = 1 - float(E_TABLE[np.searchsorted(DELTA_GRID, 0.1)].min())
print(f"deepest band absorption at 0.1 mol/m²: {100 * deepest:.1f} %")
```

## 3. The scene and the matched filter

A 64 × 64 scene: 30 channels outside methane's band (1000–2050 nm) that carry only the surface, and the 9 absorbing bands. Three surface classes in patches, with a brightness texture of a few percent. **Candidate A** is a Gaussian plume from a point source; its mask is where the truth exceeds about 500 ppm·m. **Candidate B** is a plume-shaped ellipse of a carbonate-like surface whose reflectance dips by 6 % around 2335 nm. Gaussian noise at SNR 300 per channel. Radiance is reflectance times the unit-albedo forward model, $L_b = \rho_b\, y_b(0)\, E_b(\Delta X)$, interpolated in $\Delta X$ from the table.

```python
rng = np.random.default_rng(0)
H = W = 64
WL_OUT = np.linspace(1000.0, 2050.0, 30)
WL = np.concatenate([WL_OUT, CENTERS_NM])  # (B,) = (39,)
IN = np.arange(WL_OUT.size, WL.size)

field = gaussian_filter(rng.standard_normal((H, W)), 6.0)
cls = np.digitize(field, np.quantile(field, [1 / 3, 2 / 3]))  # (H, W)
coef = np.array([[0.28, 0.06, -0.04], [0.18, 0.08, -0.04], [0.40, -0.03, -0.02]])
x_wl = (WL - 1700.0) / 700.0
rho_cls = coef @ np.stack([np.ones_like(x_wl), x_wl, x_wl**2])  # (3, B)
bright = 1 + 0.5 * gaussian_filter(rng.standard_normal((H, W)), 2.0)
rho = rho_cls[cls] * bright[..., None]  # (H, W, B)

ii, jj = np.mgrid[:H, :W]
along = jj - 14.0
sigma_y = 1.5 + 0.18 * np.clip(along, 0, None)
plume = 0.20 * np.exp(-((ii - 40.0) ** 2) / (2 * sigma_y**2)) * 1.5 / sigma_y
delta_x = np.where(along > 0, plume, 0.0)
delta_x += 0.20 * np.exp(-((ii - 40.0) ** 2 + (jj - 14.0) ** 2) / 2.0)
mask_a = delta_x > 0.02

u, v = jj - 44.0, ii - 16.0
ang = np.deg2rad(25.0)
ellipse = ((u * np.cos(ang) + v * np.sin(ang)) / 11.0) ** 2
ellipse += ((-u * np.sin(ang) + v * np.cos(ang)) / 3.2) ** 2
mask_b = ellipse < 1
dip = 1 - 0.06 * np.exp(-((WL - 2335.0) ** 2) / (2 * 15.0**2))  # (B,)

y0 = np.asarray(model(state, Lambertian(1.0), geometry))  # plume-free (B,)
E_np = np.asarray(E_TABLE)
E_pix = np.stack(
    [np.interp(delta_x.ravel(), DELTA_GRID, E_np[:, b]) for b in range(IN.size)], -1
).reshape(H, W, IN.size)
L_true = np.empty((H, W, WL.size))
L_true[..., : WL_OUT.size] = rho[..., : WL_OUT.size] * y0.mean() * (2300 / WL_OUT) ** 2
L_true[..., IN] = rho[..., IN] * y0 * E_pix
L_true[mask_b] *= dip
L = L_true * (1 + rng.standard_normal(L_true.shape) / 300.0)
masks = np.stack([mask_a, mask_b])  # (C, H, W)

# matched filter on ln L, target t (zero out of band): ΔX_MF [mol/m²]
lnL = np.log(L)
t = radtrax.unit_absorption_target(model, state, Lambertian(1.0), geometry)
target = np.concatenate([np.zeros(WL_OUT.size), np.asarray(t)])
mu = lnL.reshape(-1, WL.size).mean(0)
mf = np.asarray(apply_image(lnL, mu, estimate_cov_shrunk(lnL, mu), target))
mf_sigma = float(np.std(mf[~masks.any(0)]))
print(f"MF noise off the candidates: {mf_sigma / PPMM:,.0f} ppm·m")
for name, m in (("A", mask_a), ("B", mask_b)):
    print(
        f"{name}: {m.sum()} pixels, mean MF {mf[m].mean() / PPMM:,.0f} ppm·m, "
        f"peak {mf[m].max() / PPMM:,.0f} ppm·m, mean truth "
        f"{delta_x[m].mean() / PPMM:,.0f} ppm·m"
    )
```

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 5))
for ax, (title, im) in zip(
    axes, (("true ΔX", delta_x), ("matched filter", mf)), strict=True
):
    h = ax.imshow(im / PPMM, cmap="RdBu_r", vmin=-3000, vmax=3000)
    ax.contour(mask_a, levels=[0.5], colors="C2")
    ax.contour(mask_b, levels=[0.5], colors="C1")
    ax.set_title(title)
    ax.set_xticks([])
    ax.set_yticks([])
fig.colorbar(h, ax=axes, shrink=0.8, label="ppm·m")
axes[0].text(2, 60, "A plume", color="C2")
axes[0].text(34, 6, "B carbonate", color="C1")
plt.show()
```

The matched filter flags both, and scores the carbonate patch B higher on average than the plume A, whose mask includes the faint tail.

## 4. Vetting the two candidates

The window fits the nine absorbing bands and pairs on the 30 others; the background pool is every pixel within 16 pixels of the seeds' bounding box, outside both candidates, with a matched-filter score under twice its noise.

```python
window = VettingWindow(
    "CH4",
    fit_nm=((2290.0, 2360.0),),
    absorbing_nm=((2290.0, 2360.0),),
    mf_background_mol_m2=2 * mf_sigma,
)
results = {}
for name, m in (("A", mask_a), ("B", mask_b)):
    results[name] = vet_candidate(
        L, mf, m, WL, E_TABLE, DELTA_GRID, window, exclude=masks, box=16
    )
    r = results[name]
    print(
        f"{name}: ΔX̂ = {float(r.delta) / PPMM:,.0f} ppm·m, "
        f"D_norm = {float(r.d_norm):.2f}, {r.n_pairs} pairs"
    )
```

A fits as methane with $D_{\text{norm}}$ well below 0.3. B fits at $\widehat{\Delta X} = 0$: the continuum takes the carbonate's smooth dip, nothing methane-shaped is left, and $D_{\text{norm}} \approx 1$. A's column is below the truth over its seed pixels for two reasons: the fit returns the column of the mean transmittance, not the mean column (the transmittance is convex in $\Delta X$), and paired surfaces never match exactly.

```python
i_in, _ = pair_pixels(
    L,
    select_seeds(mf, mask_a),
    background_pool(
        mf,
        select_seeds(mf, mask_a),
        threshold=window.mf_background_mol_m2,
        exclude=masks,
        box=16,
    ),
    window.pairing_bands(WL),
)
truth_a = delta_x.ravel()[i_in].mean()
print(
    f"A: truth over its kept pixels {truth_a / PPMM:,.0f} ppm·m; the fit recovers "
    f"{100 * float(results['A'].delta) / truth_a:.0f} %"
)
fit_lin = fit_transmittance(results["A"].ratio, E_LIN, DELTA_GRID, CENTERS_NM)
print(f"A with a constant coefficient: {float(fit_lin.delta) / PPMM:,.0f} ppm·m")
```

## 5. A null from the same scene

The honest null is this scene's clutter with no plume. 100 copies of each mask, rotated and translated off both candidates, are vetted exactly as the candidate; $\hat p$ is the fraction that score at or below it.

```python
key = jax.random.key(0)
null_scores = {}
for name, m in (("A", mask_a), ("B", mask_b)):
    key, sub = jax.random.split(key)
    copies = null_masks(m, 100, key=sub, exclude=masks)
    null_scores[name] = np.array(
        [
            float(
                vet_candidate(
                    L, mf, c, WL, E_TABLE, DELTA_GRID, window, exclude=masks, box=16
                ).d_norm
            )
            for c in copies
        ]
    )
    p_hat = null_pvalue(float(results[name].d_norm), null_scores[name])
    q05, q50 = np.quantile(null_scores[name], [0.05, 0.5])
    print(
        f"{name}: p̂ = {p_hat:.2f}; null D_norm 5th percentile {q05:.2f}, "
        f"median {q50:.2f}, {int((null_scores[name] < 0.3).sum())} of 100 below 0.3"
    )
```

## Summary

- The matched filter scored the carbonate patch above the plume; the spectral fit reversed the order, accepting A and finding no methane in B.
- The transmittance comes from radtrax's forward model for the scene's own geometry and water vapour, so the curve of growth is built in; a constant coefficient biases strong plumes low.
- $D_{\text{norm}}$ thresholds depend on the window and the continuum's degree (this scene has 9 bands, EMIT's window 46), so report the scene-null $\hat p$ with every score and set the threshold per sensor on real scenes.
- Vetting's $\widehat{\Delta X}$ is a score, not a retrieval: the column for an emission rate comes from a per-pixel retrieval of the vetted plume.
